```python
from datasets import load_dataset
import pandas as pd
```

```python
ds = load_dataset('enryu43/twitter100m_users')['train']
```

```python
df = pd.DataFrame(ds)
df.to_csv('./data/twitter100m_users.csv', index=False)
top_200_users =list(df.sort_values(by='followers', ascending=False).head(200)['user'])
```

```python
# del ds
```

```python
ds = load_dataset('enryu43/twitter100m_tweets')['train'].filter(lambda x: x['user'] in top_200_users)
```

```python
df_tweets = pd.DataFrame(ds)
df_tweets.to_csv('./data/twitter100m_tweets_top200user_only.csv', index=False)
```
